```python
!pip install -q transformers sentencepiece


import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully!")
print("Device:", device)

def summarize_text(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        max_length=1024,
        truncation=True
    ).to(device)

    with torch.no_grad():
        summary_ids = model.generate(
            **inputs,
            max_new_tokens=100,
            min_new_tokens=30,
            num_beams=4,
            do_sample=False
        )

    return tokenizer.decode(
        summary_ids[0],
        skip_special_tokens=True
    )

text = """
Artificial Intelligence is transforming many industries.
It is used in healthcare, education, transportation and finance.
Machine learning allows computers to learn patterns from data.
Deep learning helps systems recognize images, understand speech
and process natural language. AI can improve productivity and
help people make better decisions.
"""

summary = summarize_text(text)

print("Original Text:\n", text)
print("\nGenerated Summary:\n", summary)
user_text = input("Enter text to summarize: ")

print("\nSummary:")
print(summarize_text(user_text))
```
