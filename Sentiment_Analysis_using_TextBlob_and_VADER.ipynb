```python

# 1. Install required libraries
!pip install -q textblob vaderSentiment

# 2. Import libraries
from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import pandas as pd

# 3. Sample Texts
texts = [
    "I love this product",
    "This movie is amazing",
    "I hate this experience",
    "The food was terrible",
    "The weather is okay",
    "I am very happy today"
]

analyzer = SentimentIntensityAnalyzer()

# 4. TextBlob Sentiment Analysis
def textblob_sentiment(text):
    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return polarity, sentiment

textblob_results = [
    textblob_sentiment(text) for text in texts
]

# 5. VADER Sentiment Analysis
def vader_sentiment(text):
    scores = analyzer.polarity_scores(text)
    compound = scores["compound"]

    if compound >= 0.05:
        sentiment = "Positive"
    elif compound <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return compound, sentiment

vader_results = [
    vader_sentiment(text) for text in texts
]

# 6. Display Results in DataFrame
results = pd.DataFrame({
    "Text": texts,
    "TextBlob Polarity": [r[0] for r in textblob_results],
    "TextBlob Sentiment": [r[1] for r in textblob_results],
    "VADER Compound": [r[0] for r in vader_results],
    "VADER Sentiment": [r[1] for r in vader_results]
})

print("Sentiment Analysis Results:")
display(results)

# 7. User Input Sentiment Analysis
user_text = input("Enter a sentence: ")

tb_score, tb_sentiment = textblob_sentiment(user_text)
vd_score, vd_sentiment = vader_sentiment(user_text)

print("\nTextBlob Result")
print("Polarity:", tb_score)
print("Sentiment:", tb_sentiment)

print("\nVADER Result")
print("Compound Score:", vd_score)
print("Sentiment:", vd_sentiment)

# 8. Compare Both Methods
print("\nComparison of TextBlob and VADER")

for i, text in enumerate(texts):
    print("\nText:", text)
    print("TextBlob:", textblob_results[i][1])
    print("VADER:", vader_results[i][1])
```
