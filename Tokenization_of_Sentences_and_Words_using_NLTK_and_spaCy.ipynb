```python
import nltk
nltk.download('punkt')
from nltk.tokenize import sent_tokenize,word_tokenize
text = "Natural Language Processing is Intresting. It is used in AI."
nltk.download('punkt_tab')
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
nltk.download('punkt_tab')
words = word_tokenize(text)
for word in words:
  print(word)
import spacy
nlp = spacy.load('en_core_web_sm')

doc = nlp(text)
for sent in doc.sents:
  print(sent.text)

for token in doc:
  print(token.text)
```
